## 1. Quick sort, utilizing a median-of-three partition

This partitioning algirithm uses the common heuristic for computing M called the median-of-three, where M is chosen as the median of the first, last and middle elements, ie. `(a[0], a[(n-1)/2], a[n-1]).`

```python
def median(a, lower, upper):
    """ 
    Inputs: an array 'a' of n integers, 'lower': the first element of the array, 'upper': the biggest element in the arra
    Goal: to compute the median of three 
    Output: returns the index of the median
    """
    middle = (lower + upper ) //2
    if (a[middle] <= a[lower] <= a[upper]) or (a[upper] <= a[lower] <= a[middle]):
        return lower
    elif (a[lower] <= a[middle] <= a[upper]) or (a[upper]<= a[middle] <= a[lower]):
        return middle
    else:
        return upper

def swap(a, i, j):
    """
    Inputs: an array 'a' of n integers, indices i, j to be swapped
    Outputs, swapped values of of i, j indices
    """
    print(f"\nSwap index {i} ({a[i]}) with index {j} ({a[j]})")
    a[i], a[j] = a[j], a[i]
    print(f"Current sorted array looks like this: {a[:]}")

def partition(a, lower, upper):
    """
    Inputs: an array 'a' of n integers, 'lower': initial index we are partitioning from, 'upper': the biggest element in the array
    Goal: partition our array with values such that place values of i,j where a[i] < = M are to its left and a[j] >= M are to its right
    Output: return the index of the newly-placed median 
    """
    #Find our median index value
    index = median(a, lower, upper) 
    #store our median
    m = a[index]

    #swap m with our final element for safe storage
    swap(a, index, upper)
    #indices for our right and left subarrays references
    i = lower
    j = upper - 1

    while i < j: 
        while i < j and a[i] < m: 
            i += 1
        while i < j and a[j] >= m:
            j -=1
        #swap if while is not satifised
        if i < j:
            swap(a, i, j)
    #our right and left subarrays have been visited
    #swap the median with our middle visited value
    swap(a, i, upper)

    #return our indice where the median was swapped to
    return i



def quicksort(a, lower, upper):
    """    
    Inputs: an array 'a' of n integers, 'lower': the first element of the array in question, 'upper': the biggest element in the array
    Goal: to have our subarray a[lower: upper] sorted
    Output: Divide-and-Conquer recursive call to itself 
    """
    #Base case
    if lower >= upper:
        return 
    
    # Two-element subarray
    if upper - lower == 1:
        if a[lower] > a[upper]:
            swap(a, lower, upper)
        return
    
    print(f"\nQuicksort({lower}, {upper})")
    print(f"Quicksort array looks like this: {a[lower:upper]}")

    #partition
    index = partition(a, lower, upper)
    print(f"Pivot ended at index {index}")

    #recursively sort LEFT side
    quicksort(a, lower, index-1)
    #Recursively sort RIGHT side
    quicksort(a, index+1, upper)

```

We will be exploring an example from class, to ensure we can see the recursive calls, swapping that occurs in our median-of-three quicksort algorithm.

```python
#array to sort: 
a = [9,6, 1, 9, 11, 10, 6, 9, 12, 2, 7]
upper = len(a)

quicksort(a, 0, upper-1)
```

## Running time of Quicksort

The running time (ie. number of steps T(n) for an array size of n comparables) of quicksort depends on how the pivot is chosen. For our previous algorithm, we decided to utilize a median-of-three pivot heuristic. 

It doesn't matter whehter the array has 10 elements or 10 million elements we are at most comparing three values, 
so we have for our median function: 
$$T_{median}(n) = \Theta(1)$$

The running time of our partition involves finding the index, and swapping which are $\Theta(1)$ time, however, the scanning part is where we are doing most of our work. Our algorithm runs nested while loops. Because the pointers are moving only in one direction, neither pointer ever resets to original position so the total work is proportional to n: 
$$T_{partition}(n) = \Theta(n)$$


The recurrence partitions the two subarrays, so that the overall recurrence is:
$$T(n) = T(k) + T(n-k) + \Theta(n)$$
where k is the number of elements on the left of the pivot

For our worst-case analysis, it would be when the pivot $a_{left}$ on the left has length equal to one a and the $a_{right}$ is a subarray respectively of n-2, n-4, n-6, down to either 1 or 2, so the running time is:

$$T(n) = \Theta(n + (n-2) + (n-4) + ... + 1 ) = \Theta( \sum_{i=1}^{(n+1)/2} (2i - 1) )= \frac{n(n+1)}{2} - 1 = \Theta (n^2)$$

and so our `Quicksort` has a worst-case quadratic running time. 

However, if our pivot divides our array roughly in half: 
$$T(n)= 2T(n/2) + cn $$

So using Master's Theorem:
$$ a = 2, b = 2, f(n) = n $$

and
$$ n^{log_{b}a}= n^{log_{2}2} = n  $$

So:
$$ T(n) = \Theta(nlogn) $$

for a balanced quicksort


## 2. Finding Order Statistic

The kth order statistic of an array a of n elements is the kth least element in the array of k = 0, ... , n - 1. Moreover, finding the kth order statistic of a can be accomplished by sorting a and returning kth element in the sorted array. Using MergeSort, this will take $ \Theta(nlogn) $ steps. We can now describe an algorithm similar to Quicksort and reduces the running time for finding the kth statistic down to $ O(n) $ steps.  

```python
def easy_find_statistic(a, lower, upper, k):
    """
    Inputs: 'a' an array of integers, lower and upper are integers, k is an index that lies between lower and upper.
    Goal: Utilized insertion sort to sort through the four elements in the array
    Output: the kth smallest index, since the array is already sorted, that will be k
    """
    #sort the small subarray using insertion sort
    print(f"    Easy statistic on indices {lower} to {upper}")
    for i in range(lower+1, upper+1):
        key = a[i]
        j = i - 1
        #Move larger elements one position to the right:
        while j>= lower and a[j] > key:
            a[j+1] = a[j]
            j -= 1
        a[j+1] = key

    print(f"    Sorted small subarray: {a[lower:upper + 1]}")
    print(f"    Returning index {k}, value {a[k]}")
    #After sorting, the kth smallest elemetn is at index k
    return k

def find_statistic(a, lower, upper, k, depth=0):
    """
    Inputs: 'a' an array of integers, lower and upper are integers, k is an index that lies between lower and upper.
    Goal: Find the kth least element in a without sorting through the entire array. 
    Output: the kth least statistic of a 
    """
    #Indentation based on recursion depth
    indent = "   " * depth
    
    print(f"{indent}Level {depth}: find_statistic({lower}, {upper})")
    print(f"{indent}Subarray: {a[lower:upper + 1]}")
    print(f"{indent}Looking for index k = {k}")


    #Base Case
    if upper - lower <= 4:
        print(f"{indent}Base case reached")
        print(f"{indent}Returning from level {depth}\n")
        return easy_find_statistic(a, lower, upper, k)
    
    
    #Recursive case:
    index = partition(a, lower, upper)
    print(f"Pivot ended at index {index}")
    if k == index:
        print(f"{indent}k == pivot index")
        print(f"{indent}Found statistic: {a[index]}")
        return index
    if k < index: 
        print(f"{indent}k < pivot index")
        print(f"{indent}Recurse LEFT\n")
        return find_statistic(a, lower, index-1, k, depth+1)
    #else k > index:
    print(f"{indent}k > pivot index")
    print(f"{indent}Recurse RIGHT\n")
    return find_statistic(a, index+1, upper, k, depth+1)
```

## Running time for Find Statistic: 

Our partition from Quicksort as solved earlier takes $\theta(n)$ time.

So the general recurrence for `find statistic` is:
$$ T(n) = T(\text{size of chosen side}) + \theta(n) $$
The runningtime depends on how large that chosen side is. 

Best cause is that the algorithm creates a partition at the exact value k. The function returns k and we do not have a recursive call.

Therefore: 
    $$ T(n) = \theta(n) $$

Worst case is if our partioning algorithm fails and our algorithm continues to produce partitions with left side length of 1, and right side where we are searching for our index k n-1, calls.

The total work then leaves us with: 
$$ T(n) = \theta(n) + \theta(n-1) + \theta(n-2) + ... + \theta(1) $$

which leads to the arithmetic sum: 
$$ T(n) = \frac {n(n+1)}{2} $$

Therefore the worst case is: 
    $$ T(n) = \theta(n^{2}) $$
